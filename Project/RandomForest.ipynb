```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
import pandas as pd

X_train = pd.read_csv("src/Dataset/X_train.csv")
X_test = pd.read_csv("src/Dataset/X_test.csv")
y_train = pd.read_csv("src/Dataset/y_train.csv")
y_test = pd.read_csv("src/Dataset/y_test.csv")

X_train = X_train.to_numpy()
X_test = X_test.to_numpy()
y_train = y_train.to_numpy().ravel()
y_test = y_test.to_numpy().ravel()
# Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Evaluate
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred, target_names=['Healthy', 'Low Risk', 'Injured']))
```
